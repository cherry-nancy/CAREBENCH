```python
import pandas as pd
```

```python
full_df = pd.read_parquet("full.parquet")
```

```python
len(full_df)
```

```python
# count number of positive samples
positive = full_df[full_df['Label'] == True]
print("Number of positive samples:", len(positive))
```

```python
# split into val and test, 70-30 ratio, save to parquet
val_df = full_df.sample(frac=0.7, random_state=42)
test_df = full_df.drop(val_df.index)
val_df.to_parquet("val.parquet", index=False)
test_df.to_parquet("test.parquet", index=False)

```

```python
len(val_df), len(test_df)
```

```python
stability = pd.read_parquet('stability.parquet')
```

```python
stability
```

```python
# split the stability dataframe into two dataframes: val_df and test_df in a 70:30 ratio
val_df = stability.sample(frac=0.7, random_state=42)
test_df = stability.drop(val_df.index)
# save the val_df and test_df dataframes to parquet files
val_df.to_parquet('stability_val.parquet', index=False)
test_df.to_parquet('stability_test.parquet', index=False)
```
