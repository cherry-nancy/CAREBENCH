```python
stability_df = pd.read_parquet("stability.parquet")
stability_df.head(10)
```

```python
# split into val and test by 70-30 ratio
val_csns = pd.read_parquet("val.parquet")["CSN"].tolist()
test_csns = pd.read_parquet("test.parquet")["CSN"].tolist()

stability_val_df = stability_df[stability_df["CSN"].isin(val_csns)]
stability_test_df = stability_df[stability_df["CSN"].isin(test_csns)]

stability_val_df.to_parquet("stability_val.parquet")
stability_test_df.to_parquet("stability_test.parquet")
print(f"Validation set size: {len(stability_val_df)}")
print(f"Test set size: {len(stability_test_df)}")
```
